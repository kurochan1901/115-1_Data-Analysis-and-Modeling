# Data Reading

## Import Dataset

```python
# Load the EastWestAirlines dataset
df = pd.read_csv("EastWestAirlines.csv")

# Preview the dataset
display(df.head())
```

## Dataset Overview

```python
# Dataset dimensions
print("Dataset shape:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset information:")
df.info()
```

## Variable Type Identification

```python
# 後續需要根據不同欄位型態採用不同計算方式
# Define variable groups
numerical_cols = [
    "Balance",
    "Qual_miles",
    "Bonus_miles",
    "Bonus_trans",
    "Flight_miles_12mo",
    "Flight_trans_12",
    "Days_since_enroll"
]

ordinal_cols = [
    "cc1_miles",
    "cc2_miles",
    "cc3_miles"
]

identifier_col = "ID"
external_col = "Award"

# Verify variable classification
print("Numerical variables:", numerical_cols)
print("Ordinal variables:", ordinal_cols)
print("Identifier:", identifier_col)
print("External variable:", external_col)

print("\nTotal classified variables:",
      len(numerical_cols) + len(ordinal_cols) + 2)
```

# Data Cleaning

## Missing Value Check and Treatment

NO Missing Value

```python
# Check missing values in each column
missing_count = df.isnull().sum()

# Calculate missing value percentage
missing_percentage = (missing_count / len(df)) * 100

# Create missing value summary
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage (%)": missing_percentage
})

display(missing_summary)

# Total missing values
print("Total missing values:", missing_count.sum())
```

## Duplicate Value Check and Treatment

```python
# Check fully duplicated rows
full_duplicates = df.duplicated().sum()

# Check duplicated member IDs
id_duplicates = df.duplicated(subset=["ID"]).sum()

# Check duplicated behavioral profiles (excluding ID)
profile_duplicates = df.duplicated(
    subset=numerical_cols + ordinal_cols
).sum()

print("Fully duplicated rows:", full_duplicates)                # 完全相同資料
print("Duplicated member IDs:", id_duplicates)                  # 重複會員ID
print("Duplicated behavioral profiles:", profile_duplicates)    # 重複行為特徵資料
```
